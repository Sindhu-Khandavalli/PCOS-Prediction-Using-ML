# MACHINE LEARNING

# **23.Feature and Target Seperation:**

```python
X = data.drop(
    columns=["PCOS_Diagnosis"]
)

y = data["PCOS_Diagnosis"]

print("INPUT FEATURES (X)")
print("=" * 70)

for number, column in enumerate(X.columns, start=1):
    print(number, ".", column)

print("\nTARGET VARIABLE (y)")
print("=" * 70)

print("PCOS_Diagnosis")

print("\nShape of X:", X.shape)
print("Shape of y:", y.shape)
```

### Description:

This section separates the independent variables from the dependent variable.
The five patient-related attributes are selected as input features.
PCOS_Diagnosis is selected as the target variable.
The input features are stored in X for model training.
The target values are stored in y for supervised classification.
This separation prepares the dataset for the train-test and modeling stages.

# **24.Train-Test Split:**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=2,
    stratify=y
)

print("TRAIN-TEST SPLIT")
print("=" * 70)

print("Total Records    :", len(data))
print("Training Records :", len(X_train))
print("Testing Records  :", len(X_test))

print("\nTraining Feature Shape:", X_train.shape)
print("Testing Feature Shape :", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

### Description:

This section divides the dataset into training and testing subsets.
The training data is used to teach the machine learning models.
The testing data is used to evaluate model performance on unseen records.
Eighty percent of the records are used for training and twenty percent for testing.
A fixed random state is used to make the experiment reproducible.
Stratification is used to maintain the target-class distribution in both subsets.

# **25.Logistic Regression:**

```python
lr = LogisticRegression(
    max_iter=1000
)

# Train the model
lr.fit(
    X_train,
    y_train
)

# Make predictions
y_pred_lr = lr.predict(
    X_test
)

# Calculate accuracy
lr_accuracy = accuracy_score(
    y_test,
    y_pred_lr
)

print("LOGISTIC REGRESSION RESULTS")
print("=" * 70)

print(
    "Accuracy:",
    round(lr_accuracy * 100, 2),
    "%"
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_lr
    )
)
```

### Description:

This section trains a Logistic Regression model for PCOS classification.
Logistic Regression is a supervised machine learning algorithm for binary classification.
The model learns relationships between the five input features and diagnosis.
The trained model is used to predict the classes of the test records.
Accuracy and a classification report are calculated to evaluate performance.
The trained model can later be saved as a pickle file for deployment.

# **26.Decision Tree:**

```python
dt = DecisionTreeClassifier(
    random_state=2
)

# Train the model
dt.fit(
    X_train,
    y_train
)

# Make predictions
y_pred_dt = dt.predict(
    X_test
)

# Calculate accuracy
dt_accuracy = accuracy_score(
    y_test,
    y_pred_dt
)

print("DECISION TREE RESULTS")
print("=" * 70)

print(
    "Accuracy:",
    round(dt_accuracy * 100, 2),
    "%"
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_dt
    )
)
```

### Description:

This section develops a Decision Tree classification model.
The Decision Tree learns decision rules from the training dataset.
These rules are used to classify patients into PCOS and non-PCOS categories.
The trained model predicts the diagnosis of the testing records.
Accuracy and classification metrics are calculated to evaluate the model.
The trained Decision Tree can also be saved for later deployment.

# **27.Support Vector machine(SVM):**

```python
svc = SVC(
    C=20,
    kernel="linear"
)

# Train the model
svc.fit(
    X_train,
    y_train
)

# Make predictions
y_pred_svc = svc.predict(
    X_test
)

# Calculate accuracy
svc_accuracy = accuracy_score(
    y_test,
    y_pred_svc
)

print("SUPPORT VECTOR MACHINE RESULTS")
print("=" * 70)

print(
    "Accuracy:",
    round(svc_accuracy * 100, 2),
    "%"
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_svc
    )
)
```

### Description:

This section trains a Support Vector Machine classification model.
SVM is a supervised learning algorithm that separates different classes using a decision boundary.
The model is trained using the five selected patient features.
A linear kernel is used for the classification task.
The trained model is evaluated using the testing dataset.
Its performance is later compared with Logistic Regression and Decision Tree.

# **28.Model Comparison:**

```python
model_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Support Vector Machine"
    ],

    "Accuracy": [
        lr_accuracy,
        dt_accuracy,
        svc_accuracy
    ]
})

model_results["Accuracy (%)"] = (
    model_results["Accuracy"] * 100
).round(2)

print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)

display(
    model_results[
        ["Model", "Accuracy (%)"]
    ]
)

# Identify best model
best_index = model_results[
    "Accuracy"
].idxmax()

best_model = model_results.loc[
    best_index,
    "Model"
]

best_accuracy = model_results.loc[
    best_index,
    "Accuracy (%)"
]

print("\nBest Performing Model")
print("=" * 70)
print("Model   :", best_model)
print("Accuracy:", best_accuracy, "%")
```

### Description:

This section compares the performance of the three trained machine learning models.
Logistic Regression, Decision Tree, and SVM are evaluated using test accuracy.
The accuracy values are converted into percentages for easier interpretation.
The results are presented together in a comparison table.
The model with the highest accuracy is identified automatically.
The selected model can be considered for deployment in the application.
